# File Handling Assignment

### 1. Counting Words in a File

```python
def count_words():
    file = open(r'C:\Users\HP\Downloads\ASSIGNMENT\data.txt', 'r')
    text = file.read()
    file.close()
    words = text.split()
    return len(words)

print('Number of words:', count_words())
```

### 2. Finding and Replacing Text

```python
def find_replace(old_word, new_word):
    file = open(r'C:\Users\HP\Downloads\ASSIGNMENT\data.txt', 'r')
    text = file.read()
    file.close()

    text = text.replace(old_word, new_word)

    file = open(r'C:\Users\HP\Downloads\ASSIGNMENT\data.txt', 'w')
    file.write(text)
    file.close()

# content before replacing
print('Before:')
file = open('data.txt', 'r')
print(file.read())
file.close()

find_replace('Python', 'Java')

# content after replacing
print('After:')
file = open(r'C:\Users\HP\Downloads\ASSIGNMENT\data.txt', 'r')
print(file.read())
file.close()
```

### 3. Reading a File in Reverse

```python
def print_reverse():
    file = open(r'C:\Users\HP\Downloads\ASSIGNMENT\data.txt', 'r')
    for line in file:
        line = line.strip()
        print(line[::-1])      # reverse the characters of each line
    file.close()

print_reverse()
```

Extra: printing the lines themselves from last line to first line

```python
def print_lines_reverse():
    file = open(r'C:\Users\HP\Downloads\ASSIGNMENT\data.txt', 'r')
    lines = file.readlines()
    file.close()
    for line in lines[::-1]:
        print(line.strip())

print_lines_reverse()
```

### 4. Counting Lines, Words and Characters

```python
def file_stats():
    file = open(r'C:\Users\HP\Downloads\ASSIGNMENT\data.txt', 'r')
    text = file.read()
    file.close()

    lines = len(text.splitlines())
    words = len(text.split())
    characters = len(text)
    return lines, words, characters

l, w, c = file_stats()
print('Lines      :', l)
print('Words      :', w)
print('Characters :', c)
```

# NumPy Assignment

### 1. Array of shape (5, 5) with random integers between 1 and 20, replace third column with 1

```python
arr = np.random.randint(1, 21, (5, 5))
print('Original array:')
print(arr)

arr[:, 2] = 1
print('After replacing third column with 1:')
print(arr)
```

### 2. Array of shape (4, 4) with values 1 to 16, replace diagonal elements with 0

```python
arr = np.arange(1, 17).reshape(4, 4)
print('Original array:')
print(arr)

np.fill_diagonal(arr, 0)
print('After replacing diagonal with 0:')
print(arr)
```

### 3. Array of shape (5, 5) with random integers, extract the border elements

```python
arr = np.random.randint(1, 50, (5, 5))
print(arr)

top = arr[0, :]
bottom = arr[-1, :]
left = arr[1:-1, 0]
right = arr[1:-1, -1]

border = np.concatenate([top, bottom, left, right])
print('Border elements:')
print(border)
```

### 4. Array of shape (4, 4) with values 1 to 16, row-wise and column-wise sum

```python
arr = np.arange(1, 17).reshape(4, 4)
print(arr)

print('Row-wise sum   :', arr.sum(axis=1))
print('Column-wise sum:', arr.sum(axis=0))
```

### 5. Array of shape (3, 3) with values 1 to 9, reshape to (1, 9) and then (9, 1)

```python
arr = np.arange(1, 10).reshape(3, 3)
print('Original (3, 3):')
print(arr)

arr1 = arr.reshape(1, 9)
print('Reshaped to (1, 9):')
print(arr1)

arr2 = arr.reshape(9, 1)
print('Reshaped to (9, 1):')
print(arr2)
```

### 6. Structured array with name, age and weight, sorted by age

```python
data = np.array([('Ravi', 25, 65.5),
                 ('Anil', 21, 70.2),
                 ('Sneha', 23, 55.0),
                 ('Kiran', 19, 60.8)],
                dtype=[('name', 'U10'), ('age', 'i4'), ('weight', 'f4')])

print('Original array:')
print(data)

sorted_data = np.sort(data, order='age')
print('Sorted by age:')
print(sorted_data)
```

### 1. DataFrame (5 rows, 3 columns) with random integers, add a column that is the product of the first two columns

```python
df = pd.DataFrame(np.random.randint(1, 20, (5, 3)), columns=['A', 'B', 'C'])
df['Product'] = df['A'] * df['B']
df
```

### 2. DataFrame (4 rows, 3 columns), row-wise and column-wise sum

```python
df = pd.DataFrame(np.random.randint(1, 20, (4, 3)), columns=['A', 'B', 'C'])
print(df)

print('Row-wise sum:')
print(df.sum(axis=1))

print('Column-wise sum:')
print(df.sum(axis=0))
```

### 3. DataFrame (5 rows, 3 columns) with NaN values, fill NaN with column mean

```python
df = pd.DataFrame(np.random.randint(1, 20, (5, 3)), columns=['A', 'B', 'C'])
df = df.astype(float)

# introduce some NaN values
df.iloc[0, 1] = np.nan
df.iloc[2, 0] = np.nan
df.iloc[3, 2] = np.nan

print('With NaN values:')
print(df)

df = df.fillna(df.mean())
print('After filling NaN with column mean:')
print(df)
```

### 4. DataFrame (6 rows, 4 columns) with NaN values, drop rows with any NaN

```python
df = pd.DataFrame(np.random.randint(1, 20, (6, 4)), columns=['A', 'B', 'C', 'D'])
df = df.astype(float)

# introduce some NaN values
df.iloc[1, 2] = np.nan
df.iloc[4, 0] = np.nan

print('With NaN values:')
print(df)

df = df.dropna()
print('After dropping rows with NaN:')
print(df)
```

### 5. Data Aggregation: group by Category and find sum and mean of Value

```python
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], 10),
    'Value': np.random.randint(1, 100, 10)
})
print(df)

df.groupby('Category')['Value'].agg(['sum', 'mean'])
```

### 6. DataFrame with Product, Category and Sales, total sales for each category

```python
df = pd.DataFrame({
    'Product': ['Laptop', 'Phone', 'Shirt', 'Jeans', 'Novel', 'Textbook', 'Tablet', 'Jacket'],
    'Category': ['Electronics', 'Electronics', 'Clothing', 'Clothing', 'Books', 'Books', 'Electronics', 'Clothing'],
    'Sales': np.random.randint(100, 1000, 8)
})
print(df)

df.groupby('Category')['Sales'].sum()
```

### 7. Merging DataFrames

```python
df1 = pd.DataFrame({'ID': [1, 2, 3, 4],
                    'Name': ['Ravi', 'Anil', 'Sneha', 'Kiran']})

df2 = pd.DataFrame({'ID': [1, 2, 3, 5],
                    'Marks': [85, 90, 78, 66]})

print(df1)
print(df2)

merged = pd.merge(df1, df2, on='ID')
merged
```

### 8. MultiIndex DataFrame with basic indexing and slicing

```python
index = pd.MultiIndex.from_tuples([('A', 1), ('A', 2), ('B', 1), ('B', 2), ('C', 1), ('C', 2)],
                                  names=['Letter', 'Number'])
df = pd.DataFrame(np.random.randint(1, 50, (6, 2)), index=index, columns=['X', 'Y'])
df
```

```python
# select all rows of letter A
df.loc['A']
```

```python
# select one row using both index levels
df.loc[('B', 2)]
```

```python
# select one single value
df.loc[('A', 1), 'X']
```

```python
# slicing from letter A to letter B
df.loc['A':'B']
```

```python
# select Number = 1 from every letter
df.xs(1, level='Number')
```

### 9. MultiIndex with Category and SubCategory, sum of values

```python
index = pd.MultiIndex.from_tuples([('Electronics', 'Phone'), ('Electronics', 'Laptop'),
                                   ('Clothing', 'Shirt'), ('Clothing', 'Jeans'),
                                   ('Electronics', 'Phone'), ('Clothing', 'Shirt')],
                                  names=['Category', 'SubCategory'])
df = pd.DataFrame({'Value': np.random.randint(10, 100, 6)}, index=index)
print(df)

print('Sum for each Category and SubCategory:')
print(df.groupby(level=['Category', 'SubCategory']).sum())
```

### 11. Apply a lambda function to create a new column that is the sum of existing columns

```python
df = pd.DataFrame(np.random.randint(1, 20, (6, 3)), columns=['A', 'B', 'C'])

df['Sum'] = df.apply(lambda row: row['A'] + row['B'] + row['C'], axis=1)
df
```
